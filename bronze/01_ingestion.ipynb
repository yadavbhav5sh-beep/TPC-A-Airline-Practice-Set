```python
%sql
-- table creations
create table if not exists sk_dev.bronze.passenger;
create table if not exists sk_dev.bronze.booking;
create table if not exists sk_dev.bronze.flight_booking;
create table if not exists sk_dev.bronze.route;
create table if not exists sk_dev.bronze.airline_route;
create table if not exists sk_dev.bronze.airline;
create table if not exists sk_dev.bronze.nation;
create table if not exists sk_dev.bronze.region;
```

```python
%sql
-- ingestion into the bronze.airline by copy into with metadata columns
COPY INTO sk_dev.bronze.airline
FROM (
  SELECT *, _metadata.file_name AS _source_file,
   current_timestamp() AS _ingestion_at,
   null as pipeline_run_id
  FROM '/Volumes/sk_dev/landing/source/airline/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```
